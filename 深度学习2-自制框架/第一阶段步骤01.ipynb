## 核心目标

整个框架的第一步，只做一件事：实现一个最简的 `Variable` 类，让它充当"存放数据的箱子"。

代码极短，但这是后面一切（`Function`、反向传播、自动微分）的基石。

## 1.1 什么是变量

用"箱子"作比喻，变量是**容器**，不是数据本身。三个要点：

| 要点 | 含义 |
| --- | --- |
| 箱子 ≠ 数据 | 箱子是容器，数据是内容 |
| 箱子里可以存放数据 | 对应「赋值」 |
| 朝箱子里看就能知道数据 | 对应「引用」，即 `x.data` |

## 1.2 实现 Variable 类

最简实现只有三行：在初始化时把传入的参数设为实例变量 `data`。

```python
class Variable:
    def __init__(self, data):
        self.data = data
```

```python
import numpy as np

x = Variable(np.array(1.0))
print(x.data)   # 1.0
```

**关键设计约定**

`Variable` 只处理 NumPy 多维数组 `numpy.ndarray` —— 因为机器学习系统的底层数据结构就是多维数组。

这里的 `x` 是"装着数据的实体"，不是数据本身。

## 1.3 补充：NumPy 多维数组

维数又称"维 / 轴"：

| 维数 | 名称 |
| --- | --- |
| 0 维 | 标量（scalar） |
| 1 维 | 向量（vector） |
| 2 维 | 矩阵（matrix） |
| N 维 | 张量（tensor） |

`ndarray.ndim` 属性返回维数。

## 这一步的"为什么"

就 3 行代码，却确立了框架最底层的抽象：**所有数据都装在 `Variable` 里，通过 `.data` 访问**。

后面的步骤会在它身上加 `grad`（梯度）、`creator`（创建者函数）等字段，逐步拼出反向传播和自动微分。
